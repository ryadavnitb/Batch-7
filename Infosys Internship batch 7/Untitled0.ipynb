```python

from google.colab import files
import io
import pandas as pd
import matplotlib.pyplot as plt

uploaded = files.upload()
df = pd.read_csv(io.BytesIO(uploaded['airlines_flights_data.csv']))


print("Shape of dataset:", df.shape)
print("\nColumns:", df.columns.tolist())
print("\nData types:")
print(df.info())
print("\nSummary statistics:")
print(df.describe())

print("\nMissing values per column:")
print(df.isnull().sum())


print("\nRows with missing prices:")
print(df[df['price'].isna()].head(4))


print("\nMedian price by airline:")
print(df.groupby('airline')['price'].median())
plt.figure(figsize=(8,5))
plt.hist(df['price'].dropna(), bins=30, edgecolor='black')
plt.title('Distribution of Flight Prices')
plt.xlabel('Price (₹)')
plt.ylabel('Frequency')
plt.show()


df['price'].fillna(df['price'].mean(), inplace=True)


print("\nAfter handling missing values:")
print(df.isnull().sum())
```

```python

import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(6,4))
sns.heatmap(df[['duration', 'days_left', 'price']].corr(), annot=True, cmap='Blues')
plt.title('Correlation Heatmap')
plt.show()



plt.figure(figsize=(10,5))
sns.boxplot(x='airline', y='price', data=df)
plt.title('Flight Price Distribution by Airline')
plt.xticks(rotation=45)
plt.show()



plt.figure(figsize=(6,4))
sns.boxplot(x='class', y='price', data=df)
plt.title('Price Comparison by Ticket Class')
plt.show()



avg_price_by_days = df.groupby('days_left')['price'].mean()

plt.figure(figsize=(8,4))
avg_price_by_days.plot(kind='line', marker='o')
plt.title('Average Flight Price vs. Days Left for Departure')
plt.xlabel('Days Left')
plt.ylabel('Average Price (₹)')
plt.grid(True)
plt.show()
```
