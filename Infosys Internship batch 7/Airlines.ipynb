```python
def optimize_memory(df):
    for col in df.columns:
        col_type = df[col].dtype

        # Convert object type columns with repeat values → category
        if col_type == 'object':
            if df[col].nunique() < df.shape[0] * 0.5:
                df[col] = df[col].astype('category')

        # Downcast integers
        elif col_type in ['int64', 'int32']:
            df[col] = pd.to_numeric(df[col], downcast='integer')

        # Downcast floats
        elif col_type in ['float64', 'float32']:
            df[col] = pd.to_numeric(df[col], downcast='float')

    return df

df = optimize_memory(df)

print("✅ Optimized Memory Usage (MB): ", df.memory_usage(deep=True).sum() / 1024**2)
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Mapping categories to approximate hours
time_map = {
    "Early_Morning": 5,
    "Morning": 9,
    "Afternoon": 14,
    "Evening": 18,
    "Night": 22,
    "Late_Night": 2
}

# Convert category → hour
df["dep_hour"] = df["departure_time"].map(time_map)

print("Mapped dep_hour values:", df["dep_hour"].notna().sum())

# Drop rows without mapping
df_clean = df.dropna(subset=["dep_hour", "duration", "source_city"])

# Aggregate for plotting
avg_df = df_clean.groupby(["source_city", "dep_hour"])["duration"].mean().reset_index()

print("Rows in aggregated df:", len(avg_df))
print(avg_df.head())

# Plot
if not avg_df.empty:
    g = sns.FacetGrid(avg_df, col="source_city", col_wrap=3, height=4)
    g.map_dataframe(sns.lineplot, x="dep_hour", y="duration", marker="o")
    g.set_axis_labels("Hour of Day", "Average Duration (hours)")
    g.set_titles("{col_name}")
    plt.show()
else:
    print("No data to plot. Check mapping or data.")
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Map labels → hour ranges
time_bin_map = {
    "Early_Morning": (4, 8),   # 4 AM – 7:59 AM
    "Morning": (8, 12),        # 8 AM – 11:59 AM
    "Afternoon": (12, 17),     # 12 PM – 4:59 PM
    "Evening": (17, 21),       # 5 PM – 8:59 PM
    "Night": (21, 24),         # 9 PM – 11:59 PM
    "Late_Night": (0, 4)       # 12 AM – 3:59 AM
}

# Create a new category column for plotting
df["time_range"] = df["departure_time"].map(lambda x: f"{time_bin_map[x][0]}–{time_bin_map[x][1]}")

print(df["time_range"].head())

# Clean data
df_clean = df.dropna(subset=["time_range", "duration", "source_city"])

# Aggregate
avg_df = df_clean.groupby(["source_city", "time_range"])["duration"].mean().reset_index()

print(avg_df.head())
```

```python
# Create time_range again
def categorize_time_range(x):
    if "Early" in x:
        return "Early_Morning"
    elif "Morning" in x:
        return "Morning"
    elif "Afternoon" in x:
        return "Afternoon"
    elif "Evening" in x:
        return "Evening"
    elif "Night" in x:
        return "Night"
    elif "Late" in x:
        return "Late_Night"
    else:
        return "Unknown"

df['time_range'] = df['departure_time'].apply(categorize_time_range)
print(df['time_range'].unique())


```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(data=df, x='days_left', y='duration', hue='time_range')
plt.title("Days Before Departure vs Duration")
plt.show()
```

```python
import matplotlib.pyplot as plt

# Calculate total delay (duration) by time of day
delay_by_time = df.groupby('time_range')['duration'].sum().sort_values()

plt.figure(figsize=(8,8))
plt.pie(delay_by_time, labels=delay_by_time.index, autopct='%1.1f%%', startangle=90)
plt.title("Delay Contribution by Time of Day")
plt.show()
```
