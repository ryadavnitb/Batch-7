```python
def optimize_memory(df):
    for col in df.columns:
        col_type = df[col].dtype

        # Convert object type columns with repeat values → category
        if col_type == 'object':
            if df[col].nunique() < df.shape[0] * 0.5:
                df[col] = df[col].astype('category')

        # Downcast integers
        elif col_type in ['int64', 'int32']:
            df[col] = pd.to_numeric(df[col], downcast='integer')

        # Downcast floats
        elif col_type in ['float64', 'float32']:
            df[col] = pd.to_numeric(df[col], downcast='float')

    return df

df = optimize_memory(df)

print("✅ Optimized Memory Usage (MB): ", df.memory_usage(deep=True).sum() / 1024**2)
```

```python
sample_fraction = 0.10  # change to 0.20 for 20%, etc.

df_sample = df.sample(frac=sample_fraction, random_state=42)

print(f"📌 Sample Created: {df_sample.shape[0]} rows selected out of {df.shape[0]}")
```

```python
print("Original Shape:", df.shape)
print("Sampled Shape:", df_sample.shape)

print("\nColumn Comparison:")
print(df.describe(include='all'))
print(df_sample.describe(include='all'))
```

```python
categorical_column = df.select_dtypes(include=['category']).columns.tolist()

if categorical_column:
    print("\n Checking sampling balance for categorical columns:")
    for col in categorical_column:
        print(f"\nCategory distribution in {col}:")
        print("Original:")
        print(df[col].value_counts(normalize=True) * 100)

        print("\nSample:")
        print(df_sample[col].value_counts(normalize=True) * 100)
```

```python
import pandas as pd
import datetime


# ---------------------------
# 1. Format and clean datetime
# ---------------------------

# Extract date part from 'departure_time' if it contains actual dates
df['departure_date'] = df['departure_time'].str.extract(r'(\d{4}-\d{2}-\d{2})')
df['departure_date'] = pd.to_datetime(df['departure_date'], errors='coerce')  # Convert to datetime
df['departure_date'] = df['departure_date'].dt.strftime('%Y-%m-%d')  # Standard YYYY-MM-DD format

# ---------------------------
# 2. Derived features from datetime
# ---------------------------
df['Month'] = pd.to_datetime(df['departure_date'], errors='coerce').dt.month
df['Day_of_Week'] = pd.to_datetime(df['departure_date'], errors='coerce').dt.day_name()

# ---------------------------
# 3. Map departure_time text to Hour
# ---------------------------
time_mapping = {
    "Early Morning": 6,
    "Morning": 9,
    "Afternoon": 14,
    "Evening": 18,
    "Night": 21
}
df['Hour'] = df['departure_time'].map(time_mapping)

# ---------------------------
# 4. Create Route feature
# ---------------------------
df['Route'] = df['source_city'] + " → " + df['destination_city']

# ---------------------------
# 5. Check final DataFrame
# ---------------------------
print(df[['departure_time','departure_date','Month','Day_of_Week','Hour','Route']].head())
```
