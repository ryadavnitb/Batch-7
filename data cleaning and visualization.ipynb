```python
import pandas as pd

df = pd.read_csv("flights_sample_3m.csv", low_memory=False)
```

```python
df_sample = df.sample(500000, random_state=42)
```

```python
df["ORIGIN_CITY"] = df["ORIGIN_CITY"].astype("category")

```

```python
df["DEST_CITY"] = df["DEST_CITY"].astype("category")
```

```python
df["DEP_DELAY"] = df["DEP_DELAY"].fillna(df["DEP_DELAY"].median())

```

```python
df["ARR_DELAY"] = df["ARR_DELAY"].fillna(df["ARR_DELAY"].median())
```

```python
df["CANCELLED"] = df["CANCELLED"].fillna(0)


```

```python
df.isnull().sum()

```

```python
df["FL_DATE"] = pd.to_datetime(df["FL_DATE"])
```

```python
df["DEP_TIME"] = df["DEP_TIME"].astype("float").astype("Int64")
```

```python
df["DEP_TIME"] = df["DEP_TIME"].astype(str).str.zfill(4)
```

```python

df["DEP_TIME"].head()
```

```python
df["DEP_DT"] = pd.to_datetime(df["DEP_TIME"], format="%H%M", errors="coerce")
```

```python
df["DEP_DT"].head()
```

```python
df["MONTH"] = df["FL_DATE"].dt.month
```

```python
df["DAY_OF_WEEK"] = df["FL_DATE"].dt.dayofweek
```

```python
df["DEP_HOUR"] = df["DEP_DT"].dt.hour
df["DEP_HOUR"].head()
```

```python
df["DEP_TIME"].head()



```

```python
df["ROUTE"] = df["ORIGIN_CITY"].str.cat(df["DEST_CITY"], sep="-")

```

```python
df["DEP_TIME"].head(20)

```

```python
df["DEP_HOUR"].isnull().sum()
```

```python

df.to_csv("cleaned_flights_sample_3m.csv", index=False)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(10,5))
sns.histplot(df["DEP_DELAY"], bins=50, kde=True)
plt.title("Distribution of Departure Delays")
plt.xlabel("Departure Delay (minutes)")
plt.ylabel("Number of Flights")
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.boxplot(x=df["DEP_DELAY"])
plt.title("Outliers in Departure Delay")
plt.show()
```

```python
plt.figure(figsize=(12,5))
sns.barplot(x="MONTH", y="DEP_DELAY", data=df)
plt.title("Average Departure Delay by Month")
plt.show()
```

```python
#Average Delay by Month
plt.figure(figsize=(12,5))
sns.barplot(x="MONTH", y="DEP_DELAY", data=df)
plt.title("Average Departure Delay by Month")
plt.show()
```

```python
#Delay by Hour of Day
plt.figure(figsize=(12,5))
sns.lineplot(x="DEP_HOUR", y="DEP_DELAY", data=df)
plt.title("Delays by Hour of Day")
plt.show()
```

```python
#top delayed airports
df_airport = df.groupby("ORIGIN_CITY")["DEP_DELAY"].mean().sort_values(ascending=False).head(10)

plt.figure(figsize=(12,5))
sns.barplot(x=df_airport.index, y=df_airport.values)
plt.title("Top 10 Most Delayed Origin Airports")
plt.xticks(rotation=45)
plt.show()
```

```python
#Airline Delay Comparison
plt.figure(figsize=(12,5))
sns.boxplot(x="AIRLINE", y="DEP_DELAY", data=df)
plt.title("Delay Comparison Across Airlines")
plt.show()
```

```python
#Most Delayed Routes
df_route = df.groupby("ROUTE")["DEP_DELAY"].mean().sort_values(ascending=False).head(10)

plt.figure(figsize=(12,5))
sns.barplot(x=df_route.index, y=df_route.values)
plt.title("Top 10 Worst Routes by Delay")
plt.xticks(rotation=90)
plt.show()
```

```python
#Delay by Day of Week vs Hour
df_pivot = df.pivot_table(
    values="DEP_DELAY",
    index="DAY_OF_WEEK",
    columns="DEP_HOUR",
    aggfunc="mean"
)

plt.figure(figsize=(14,6))
sns.heatmap(df_pivot, cmap="coolwarm")
plt.title("Delay Heatmap: Day of Week vs Hour of Day")
plt.show()
```
