```python
# 1. Merge df_flights with df_airlines to get airlines name
df = pd.merge(
    df_flights,
    df_airlines,
    left_on="AIRLINE",
    right_on="IATA_CODE",
    how="left",
    suffixes=('','_NAME')
)

if 'IATA_CODE' in df.columns:
    df.drop('IATA_CODE', axis=1, inplace=True)

# 2. Merge with df_airports for Origin Airport information
df = pd.merge(
        df,
        df_airports,
        left_on='ORIGIN_AIRPORT',
        right_on='IATA_CODE',
        how='left',
        suffixes=('', '_ORIGIN')
    )
df.rename(columns={
        'AIRPORT': 'ORIGIN_AIRPORT_NAME',
        'CITY': 'ORIGIN_CITY',
        'STATE': 'ORIGIN_STATE',
        'LATITUDE': 'ORIGIN_LATITUDE',
        'LONGITUDE': 'ORIGIN_LONGITUDE',
        'COUNTRY': 'ORIGIN_COUNTRY'
    }, inplace=True)

if 'IATA_CODE' in df:
  df.drop('IATA_CODE', axis=1, inplace=True)


# 3. Merge with df_airports for Destination Airport information
df = pd.merge(
        df,
        df_airports,
        left_on='DESTINATION_AIRPORT',
        right_on='IATA_CODE',
        how='left',
        suffixes=('', '_DESTINATION')
)
df = df.rename(columns={
        'AIRPORT': 'DESTINATION_AIRPORT_NAME',
        'CITY': 'DESTINATION_CITY',
        'STATE': 'DESTINATION_STATE',
        'LATITUDE': 'DESTINATION_LATITUDE',
        'LONGITUDE': 'DESTINATION_LONGITUDE',
        'COUNTRY': 'DESTINATION_COUNTRY'
    })
if 'IATA_CODE' in df:
  df.drop('IATA_CODE', axis=1, inplace=True)
```

Handling Missing Values

```python
# Fill delay reason columns (missing = no delay)
delay_cols = ["AIR_SYSTEM_DELAY", "SECURITY_DELAY", "AIRLINE_DELAY",
              "LATE_AIRCRAFT_DELAY", "WEATHER_DELAY"]
df[delay_cols] = df[delay_cols].fillna(0)

# Fill cancellation reason
df["CANCELLATION_REASON"] = df["CANCELLATION_REASON"].fillna("None")

# Time columns
time_cols = ["DEPARTURE_TIME", "DEPARTURE_DELAY", "TAXI_OUT", "WHEELS_OFF",
             "ELAPSED_TIME", "AIR_TIME", "WHEELS_ON", "TAXI_IN",
             "ARRIVAL_TIME", "ARRIVAL_DELAY"]

# For cancelled flights → fill with 0
df.loc[df["CANCELLED"] == 1, time_cols] = df.loc[df["CANCELLED"] == 1, time_cols].fillna(0)

# For non-cancelled flights → numeric imputation using median
for col in time_cols:
    median_val = df.loc[df["CANCELLED"] == 0, col].median()
    df.loc[(df["CANCELLED"] == 0) & (df[col].isna()), col] = median_val

# Location categorical columns → mode
loc_cols = [
    "ORIGIN_AIRPORT_NAME", "ORIGIN_CITY", "ORIGIN_STATE", "ORIGIN_COUNTRY",
    "DESTINATION_AIRPORT_NAME", "DESTINATION_CITY", "DESTINATION_STATE", "DESTINATION_COUNTRY"
]
for col in loc_cols:
    mode_val = df[col].mode()[0]
    df[col] = df[col].fillna(mode_val)

# Lat/Long numeric → median
latlong_cols = ["ORIGIN_LATITUDE", "ORIGIN_LONGITUDE",
                "DESTINATION_LATITUDE", "DESTINATION_LONGITUDE"]
for col in latlong_cols:
    df[col] = df[col].fillna(df[col].median())

# TAIL_NUMBER — keep or drop based on your requirement
df["TAIL_NUMBER"] = df["TAIL_NUMBER"].fillna(df["TAIL_NUMBER"].mode()[0])   # OR → df.drop(columns=["TAIL_NUMBER"], inplace=True)
```

Memory Optimization

```python
def optimize_memory(df):
  initial_mem = df.memory_usage().sum() / 1024**2
  print(f"Initial Memory Usage: {initial_mem:.2f} MB")

  for col in df.columns:
    col_type = df[col].dtype

    # Numeric - Integer
    if col_type in ["int64"]:
      df[col] = pd.to_numeric(df[col],downcast='integer')

    # Numeric - Floats
    elif col_type == 'float64':
        df[col] = pd.to_numeric(df[col], downcast='float')

    # Boolean style numeric columns (0/1)
    elif col in ["DIVERTED", "CANCELLED"]:
        df[col] = df[col].astype('int8')

    # Strings with repeated values → convert to category
    elif col_type == 'object':
        num_unique = df[col].nunique()
        num_total = len(df[col])
        if num_unique / num_total < 0.5:      # high repetition
            df[col] = df[col].astype('category')

  end_mem = df.memory_usage().sum() / 1024**2
  print(f"Memory usage after: {end_mem:.2f} MB")
  print(f"Reduced by: {100 * (initial_mem - end_mem) / initial_mem:.2f}%")

  return df
```

```python
df = optimize_memory(df)
```

# Feature Engineering

Time-based Features

```python
df["FLIGHT_DATE"] = pd.to_datetime(df[["YEAR", "MONTH", "DAY"]])

df["IS_WEEKEND"] = (df["DAY_OF_WEEK"].isin([6, 7])).astype(int)
df["IS_PEAK_MONTH"] = df["MONTH"].isin([6,7,8,11,12]).astype(int)  # Holidays

df["DEPARTURE_HOUR"] = (df["SCHEDULED_DEPARTURE"] // 100).astype(int)
df["ARRIVAL_HOUR"] = (df["SCHEDULED_ARRIVAL"] // 100).astype(int)
```

Weather / Airport Congestion Proxy Features

```python
df["IS_LONG_ROUTE"] = (df["DISTANCE"] > 1500).astype(int)
df["BUSY_ORIGIN"] = df["ORIGIN_AIRPORT"].map(df["ORIGIN_AIRPORT"].value_counts())
df["BUSY_DESTINATION"] = df["DESTINATION_AIRPORT"].map(df["DESTINATION_AIRPORT"].value_counts())
df["AIRLINE_TRAFFIC"] = df["AIRLINE"].map(df["AIRLINE"].value_counts())
```

Delay Propagation Features

```python
df["TOTAL_DELAY_CAUSE"] = (
    df["AIR_SYSTEM_DELAY"] + df["SECURITY_DELAY"] + df["AIRLINE_DELAY"] +
    df["LATE_AIRCRAFT_DELAY"] + df["WEATHER_DELAY"]
)

df["HAS_DELAY_CAUSE"] = (df["TOTAL_DELAY_CAUSE"] > 0).astype(int)
```

Flights that were late previously tend to be late again.

```python
df = df.sort_values(["TAIL_NUMBER", "FLIGHT_DATE"])
df["PREVIOUS_DELAY"] = df.groupby("TAIL_NUMBER")["ARRIVAL_DELAY"].shift(1).fillna(0)
df["PREVIOUS_DELAY_FLAG"] = (df["PREVIOUS_DELAY"] > 0).astype(int)
```

Target Variable Engineering

```python
df["ARRIVED_LATE"] = (df["ARRIVAL_DELAY"] > 15).astype(int)
```

Categorical Encoding

```python
cat_cols = ["AIRLINE", "ORIGIN_AIRPORT", "DESTINATION_AIRPORT"]
df = pd.get_dummies(df, columns=cat_cols, drop_first=True)   # one-hot encoding
```

```python
numerical_cols = df.select_dtypes(include=np.number).columns

# Columns to be excluded from the plot
columns_to_exclude = [
    'YEAR', 'MONTH', 'DAY', 'FLIGHT_NUMBER', 'DIVERTED', 'CANCELLED',
    'ORIGIN_LATITUDE', 'ORIGIN_LONGITUDE', 'DESTINATION_LATITUDE',
    'DESTINATION_LONGITUDE', 'IS_WEEKEND', 'IS_PEAK_MONTH', 'IS_LONG_ROUTE',
    'BUSY_ORIGIN', 'BUSY_DESTINATION', 'AIRLINE_TRAFFIC', 'FLIGHT_DATE', 'DEPARTURE_HOUR', 'ARRIVAL_HOUR', 'ARRIVED_LATE', 'TOTAL_DELAY_CAUSE', 'HAS_DELAY_CAUSE', 'PREVIOUS_DELAY', 'PREVIOUS_DELAY_FLAG'
]

# Final columns to check for outliers
cols_for_outlier = [col for col in numerical_cols if col not in columns_to_exclude]
```

```python
def plot_outliers(data, title):
    n_cols = 4
    n_rows = (len(cols_for_outlier) + n_cols - 1) // n_cols

    plt.figure(figsize=(n_cols * 5, n_rows * 4))
    plot_idx = 1

    for col in cols_for_outlier:
        plt.subplot(n_rows, n_cols, plot_idx)
        sns.boxplot(y=data[col],
                    flierprops={"marker": ".", "markersize": 3})  # smaller dots
        plt.title(col)
        plt.ylabel('')
        plot_idx += 1

    plt.suptitle(title, fontsize=18, y=1.02)
    plt.tight_layout()
    plt.show()


plot_outliers(df, "Before Outlier Removal")
```

```python
chunks = pd.read_csv("dataset/flights.csv", chunksize=300000)
clean_chunks = []

for chunk in chunks:
    for col in cols_for_outlier:
        Q1 = chunk[col].quantile(0.25)
        Q3 = chunk[col].quantile(0.75)
        IQR = Q3 - Q1
        lower = Q1 - 1.5 * IQR
        upper = Q3 + 1.5 * IQR
        chunk = chunk[(chunk[col] >= lower) & (chunk[col] <= upper)]
    clean_chunks.append(chunk)

df_clean = pd.concat(clean_chunks, ignore_index=True)
```
