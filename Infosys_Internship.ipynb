Handling Missing Values

```python
# Fill delay reason columns (missing = no delay)
delay_cols = ["AIR_SYSTEM_DELAY", "SECURITY_DELAY", "AIRLINE_DELAY",
              "LATE_AIRCRAFT_DELAY", "WEATHER_DELAY"]
df[delay_cols] = df[delay_cols].fillna(0)

# Fill cancellation reason
df["CANCELLATION_REASON"] = df["CANCELLATION_REASON"].fillna("None")

# Time columns
time_cols = ["DEPARTURE_TIME", "DEPARTURE_DELAY", "TAXI_OUT", "WHEELS_OFF",
             "ELAPSED_TIME", "AIR_TIME", "WHEELS_ON", "TAXI_IN",
             "ARRIVAL_TIME", "ARRIVAL_DELAY"]

# For cancelled flights → fill with 0
df.loc[df["CANCELLED"] == 1, time_cols] = df.loc[df["CANCELLED"] == 1, time_cols].fillna(0)

# For non-cancelled flights → numeric imputation using median
for col in time_cols:
    median_val = df.loc[df["CANCELLED"] == 0, col].median()
    df.loc[(df["CANCELLED"] == 0) & (df[col].isna()), col] = median_val

# Location categorical columns → mode
loc_cols = [
    "ORIGIN_AIRPORT_NAME", "ORIGIN_CITY", "ORIGIN_STATE", "ORIGIN_COUNTRY",
    "DESTINATION_AIRPORT_NAME", "DESTINATION_CITY", "DESTINATION_STATE", "DESTINATION_COUNTRY"
]
for col in loc_cols:
    mode_val = df[col].mode()[0]
    df[col] = df[col].fillna(mode_val)

# Lat/Long numeric → median
latlong_cols = ["ORIGIN_LATITUDE", "ORIGIN_LONGITUDE",
                "DESTINATION_LATITUDE", "DESTINATION_LONGITUDE"]
for col in latlong_cols:
    df[col] = df[col].fillna(df[col].median())

# TAIL_NUMBER — keep or drop based on your requirement
df["TAIL_NUMBER"] = df["TAIL_NUMBER"].fillna(df["TAIL_NUMBER"].mode()[0])   # OR → df.drop(columns=["TAIL_NUMBER"], inplace=True)
```

Memory Optimization

```python
def optimize_memory(df):
  initial_mem = df.memory_usage().sum() / 1024**2
  print(f"Initial Memory Usage: {initial_mem:.2f} MB")

  for col in df.columns:
    col_type = df[col].dtype

    # Numeric - Integer
    if col_type in ["int64"]:
      df[col] = pd.to_numeric(df[col],downcast='integer')

    # Numeric - Floats
    elif col_type == 'float64':
        df[col] = pd.to_numeric(df[col], downcast='float')

    # Boolean style numeric columns (0/1)
    elif col in ["DIVERTED", "CANCELLED"]:
        df[col] = df[col].astype('int8')

    # Strings with repeated values → convert to category
    elif col_type == 'object':
        num_unique = df[col].nunique()
        num_total = len(df[col])
        if num_unique / num_total < 0.5:      # high repetition
            df[col] = df[col].astype('category')

  end_mem = df.memory_usage().sum() / 1024**2
  print(f"Memory usage after: {end_mem:.2f} MB")
  print(f"Reduced by: {100 * (initial_mem - end_mem) / initial_mem:.2f}%")

  return df
```

```python
df = optimize_memory(df)
```

# **Week2: Data Preprocessing and Feature Engineering**

# Feature Engineering

Time-based Features

```python
df["FLIGHT_DATE"] = pd.to_datetime(df[["YEAR", "MONTH", "DAY"]])

df["IS_WEEKEND"] = (df["DAY_OF_WEEK"].isin([6, 7])).astype(int)
df["IS_PEAK_MONTH"] = df["MONTH"].isin([6,7,8,11,12]).astype(int)  # Holidays

df["DEPARTURE_HOUR"] = (df["SCHEDULED_DEPARTURE"] // 100).astype(int)
df["ARRIVAL_HOUR"] = (df["SCHEDULED_ARRIVAL"] // 100).astype(int)
```

Weather / Airport Congestion Proxy Features

```python
df["IS_LONG_ROUTE"] = (df["DISTANCE"] > 1500).astype(int)
df["BUSY_ORIGIN"] = df["ORIGIN_AIRPORT"].map(df["ORIGIN_AIRPORT"].value_counts())
df["BUSY_DESTINATION"] = df["DESTINATION_AIRPORT"].map(df["DESTINATION_AIRPORT"].value_counts())
df["AIRLINE_TRAFFIC"] = df["AIRLINE"].map(df["AIRLINE"].value_counts())
```

Delay Propagation Features

```python
df["TOTAL_DELAY_CAUSE"] = (
    df["AIR_SYSTEM_DELAY"] + df["SECURITY_DELAY"] + df["AIRLINE_DELAY"] +
    df["LATE_AIRCRAFT_DELAY"] + df["WEATHER_DELAY"]
)

df["HAS_DELAY_CAUSE"] = (df["TOTAL_DELAY_CAUSE"] > 0).astype(int)
```

Flights that were late previously tend to be late again.

```python
df = df.sort_values(["TAIL_NUMBER", "FLIGHT_DATE"])
df["PREVIOUS_DELAY"] = df.groupby("TAIL_NUMBER")["ARRIVAL_DELAY"].shift(1).fillna(0)
df["PREVIOUS_DELAY_FLAG"] = (df["PREVIOUS_DELAY"] > 0).astype(int)
```

Target Variable Engineering

```python
df["ARRIVED_LATE"] = (df["ARRIVAL_DELAY"] > 15).astype(int)
```

Categorical Encoding

```python
cat_cols = ["AIRLINE", "ORIGIN_AIRPORT", "DESTINATION_AIRPORT"]
df = pd.get_dummies(df, columns=cat_cols, drop_first=True)   # one-hot encoding
```

```python
def plot_outliers(data, title):
    n_cols = 4
    n_rows = (len(cols_for_outlier) + n_cols - 1) // n_cols

    plt.figure(figsize=(n_cols * 5, n_rows * 4))
    plot_idx = 1

    for col in cols_for_outlier:
        plt.subplot(n_rows, n_cols, plot_idx)
        sns.boxplot(y=data[col],
                    flierprops={"marker": ".", "markersize": 3})  # smaller dots
        plt.title(col)
        plt.ylabel('')
        plot_idx += 1

    plt.suptitle(title, fontsize=18, y=1.02)
    plt.tight_layout()
    plt.show()


plot_outliers(df, "Before Outlier Removal")
```

5.1 Top 10 Origin-Destination Pairs

```python
top_routes = df.groupby(['ORIGIN_AIRPORT_NAME', 'DESTINATION_AIRPORT_NAME'], observed=False).agg({
    'FLIGHT_NUMBER': 'count',
    'DEPARTURE_DELAY': 'mean',
    'ARRIVAL_DELAY': 'mean'
}).round(2).reset_index()

top_routes.columns = ['Origin', 'Destination', 'Total_Flights', 'Avg_Dep_Delay', 'Avg_Arr_Delay']
top_10_routes = top_routes.nlargest(10, 'Total_Flights')

# Convert categorical to string
top_10_routes['Origin'] = top_10_routes['Origin'].astype(str)
top_10_routes['Destination'] = top_10_routes['Destination'].astype(str)

# SHORTEN NAMES + ADD ELLIPSIS
def shorten_name(name, max_len=18):
    if len(name) > max_len:
        return name[:max_len-3] + '...'
    return name

top_10_routes['Short_Origin'] = top_10_routes['Origin'].apply(shorten_name)
top_10_routes['Short_Destination'] = top_10_routes['Destination'].apply(shorten_name)
top_10_routes['Route'] = top_10_routes['Short_Origin'] + ' → ' + top_10_routes['Short_Destination']

plt.figure(figsize=(8, 5))
colors = plt.cm.RdYlGn_r(np.linspace(0.3, 1, len(top_10_routes)))

bars = plt.barh(top_10_routes['Route'], top_10_routes['Total_Flights'], color=colors, alpha=0.8)
plt.xlabel('Total Flights', fontsize=12, fontweight='bold')
plt.title('Top 10 Busiest Routes by Flight Volume', fontsize=16, fontweight='bold', pad=20)
plt.gca().invert_yaxis()

plt.tight_layout()
plt.show()

print("✅ Top 10 Routes - Clean labels with ellipsis!")
print(top_10_routes[['Route', 'Total_Flights', 'Avg_Arr_Delay']].round(1))
```

```python
monthly_cancel = df.groupby('MONTH').agg({
    'CANCELLED': ['count', 'sum'],
    'ARRIVAL_DELAY': 'mean'
}).round(2)
monthly_cancel.columns = ['Total_Flights', 'Cancellations', 'Avg_Delay']
monthly_cancel['Cancel_Rate'] = (monthly_cancel['Cancellations'] / monthly_cancel['Total_Flights'] * 100)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 6))

month_names = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 
               'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
x = range(1, 13)

ax1.bar(x, monthly_cancel['Cancel_Rate'], color='crimson', alpha=0.7)
ax1.set_title('Monthly Cancellation Rate')
ax1.set_xlabel('Month')
ax1.set_ylabel('Cancellation Rate (%)')
ax1.set_xticks(x)
ax1.set_xticklabels(month_names)

ax2.plot(x, monthly_cancel['Avg_Delay'], 'o-', color='orange', linewidth=2, markersize=8)
ax2.set_title('Monthly Average Delay')
ax2.set_xlabel('Month')
ax2.set_ylabel('Avg Delay (min)')
ax2.set_xticks(x)
ax2.set_xticklabels(month_names)
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

6.3 Holiday/Winter Impact Analysis

```python
# Define holiday periods
holidays = [(11,23,12,1), (12,20,12,31), (6,28,7,4)]
df['HOLIDAY_PERIOD'] = 0
for start_mon, start_day, end_mon, end_day in holidays:
    mask = ((df['MONTH'] == start_mon) & (df['DAY'] >= start_day)) | \
           ((df['MONTH'] == end_mon) & (df['DAY'] <= end_day))
    df.loc[mask, 'HOLIDAY_PERIOD'] = 1

# Seasonal comparison
winter_months = [12,1,2]
summer_months = [6,7,8]

winter_cancel = df[df['MONTH'].isin(winter_months)]['CANCELLED'].mean() * 100
summer_cancel = df[df['MONTH'].isin(summer_months)]['CANCELLED'].mean() * 100

plt.figure(figsize=(7, 4))
seasons = ['Winter (Dec-Feb)', 'Summer (Jun-Aug)', 'Holiday Periods']
rates = [winter_cancel, summer_cancel, df[df['HOLIDAY_PERIOD']==1]['CANCELLED'].mean()*100]
colors = ['#2E86C1', '#A23B72', '#DC7633']

bars = plt.bar(seasons, rates, color=colors, alpha=0.8, edgecolor='black')
plt.title('Cancellation Rates: Seasonal & Holiday Comparison')
plt.ylabel('Cancellation Rate (%)')
plt.ylim(0, max(rates)*1.2)

for bar, rate in zip(bars, rates):
    plt.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.1, 
             f'{rate:.2f}%', ha='center', va='bottom', fontweight='bold')

plt.xticks(rotation=15)
plt.tight_layout()
plt.show()
```
