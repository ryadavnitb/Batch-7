1. Route-Level Price Analysis (Comparing Prices Across Routes)
This answers: Which routes are the most expensive, and how does the price distribution look

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Assuming your DataFrame is loaded as 'df'
df = pd.read_csv("airlines_flights_data.csv")
df['duration'] = pd.to_numeric(df['duration'], errors='coerce')
df.dropna(subset=['duration'], inplace=True)
df['route'] = df['source_city'] + ' -> ' + df['destination_city']

print("--- 1. Route-Level Price Distribution Analysis ---")

# Identify the top 10 busiest routes
top_10_routes = df['route'].value_counts().nlargest(10).index
df_top_routes = df[df['route'].isin(top_10_routes)]

plt.figure(figsize=(14, 8))
# Boxplot for price distribution across top routes
sns.boxplot(x='price', y='route', data=df_top_routes,
            order=df_top_routes.groupby('route')['price'].median().sort_values(ascending=False).index,
            palette='RdYlGn_r', showfliers=False) # showfliers=False to keep the plot clean
plt.title('Price Distribution for Top 10 Busiest Routes (Boxplot)')
plt.xlabel('Price (INR)')
plt.ylabel('Route (Sorted by Median Price)')
plt.tight_layout()
plt.show()
```

2. Airport-Level Price vs. Duration Analysis (Source City)
This answers: Do flights departing from certain cities cost more or have longer durations, regardless of the destination

```python
# @title Default title text
print("\n--- 2. Airport-Level Price and Duration Comparison ---")

# Group by source city and calculate the median price and median duration
city_metrics = df.groupby('source_city').agg(
    median_price=('price', 'median'),
    median_duration=('duration', 'median')
).reset_index()

# Create a figure with two subplots side-by-side
fig, axes = plt.subplots(1, 2, figsize=(15, 6))

# Subplot 1: Median Price by Source City
sns.barplot(x='median_price', y='source_city', data=city_metrics.sort_values('median_price', ascending=False),
            ax=axes[0], palette='Blues_d')
axes[0].set_title('Median Price by Source City')
axes[0].set_xlabel('Median Price (INR)')
axes[0].set_ylabel('Source City')

# Subplot 2: Median Duration by Source City
sns.barplot(x='median_duration', y='source_city', data=city_metrics.sort_values('median_duration', ascending=False),
            ax=axes[1], palette='Oranges_d')
axes[1].set_title('Median Duration by Source City')
axes[1].set_xlabel('Median Duration (Hours)')
axes[1].set_ylabel('') # Keep the y-label only on the first plot for clarity

plt.tight_layout()
plt.show()
```
