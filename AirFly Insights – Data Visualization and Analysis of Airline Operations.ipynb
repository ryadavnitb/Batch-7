```python
# Replace delay-related missing values with 0
delay_cols = ['CarrierDelay', 'WeatherDelay', 'NASDelay', 'SecurityDelay', 'LateAircraftDelay']
for col in delay_cols:
    if col in data.columns:
        data[col] = data[col].fillna(0)
```

```python
data.info()
```

```python
# Replace missing cancellation code with 'None'
if 'CancellationCode' in data.columns:
    data['CancellationCode'] = data['CancellationCode'].fillna('None')
```

```python
data.info()
```

```python
# Replace missing values in numerical columns with median
num_cols = data.select_dtypes(include=['float64','int64']).columns
for col in num_cols:
    data[col] = data[col].fillna(data[col].median())
```

```python
data.info()
```

```python
# Replace missing values in string/object columns with mode
obj_cols = data.select_dtypes(include=['object']).columns
for col in obj_cols:
    data[col] = data[col].fillna(data[col].mode()[0])
```

```python
#Feature Engineering (Month, Day, Hour, Route)


#Convert date column to datetime



data['FL_DATE'] = pd.to_datetime(data[['YEAR', 'MONTH', 'DAY']])

```

```python
print(data.head())
```

```python
#Convert Day Number to Day Name

data['DAY_NAME'] = data['FL_DATE'].dt.day_name()
```

```python
#Extract Hour from Scheduled Departure

data['DEP_HOUR'] = (data['SCHEDULED_DEPARTURE'] // 100).astype('Int64')
```

```python
#Create Route Column
data['ROUTE'] = data['ORIGIN_AIRPORT'] + "-" + data['DESTINATION_AIRPORT']
```

```python
#Check Output
data[['FL_DATE','DAY_NAME','DEP_HOUR','ROUTE']].head()
```

```python
#Handling Missing Values for This Dataset
delay_cols = ['AIR_SYSTEM_DELAY', 'SECURITY_DELAY', 'AIRLINE_DELAY', 'LATE_AIRCRAFT_DELAY', 'WEATHER_DELAY']

for col in delay_cols:
    data[col] = data[col].fillna(0)
```

```python
#Check Null Percentages
null_percent = (data.isnull().sum() / len(data)) * 100
print(null_percent)
```

```python
#Summary for Documentation
   Created a new date field (FL_DATE) by combining YEAR, MONTH, and DAY columns.
Converted numeric day value to day name to support trend analysis.
Derived DEP_HOUR from SCHEDULED_DEPARTURE column for time-based insights.
Created ROUTE column to enable route popularity and congestion analysis.
Treated missing values in delay-related fields by filling with zero.
```

```python
#Visualization 1: Top Airlines by Total Flights

import matplotlib.pyplot as plt

data['AIRLINE'].value_counts().plot(kind='bar', figsize=(10,5))
plt.title("Top Airlines by Flight Count")
plt.xlabel("Airlines")
plt.ylabel("Number of Flights")
plt.show()
```

```python
#Visualization 2: Busiest Airports
data['ORIGIN_AIRPORT'].value_counts().head(10).plot(kind='bar', figsize=(10,5))
plt.title("Top 10 Busiest Origin Airports")
plt.xlabel("Airport")
plt.ylabel("Flights")
plt.show()
```

```python
#3. Line Plot – Monthly Flight Trend
data['MONTH'].value_counts().sort_index().plot(kind='line', figsize=(10,5))
plt.title("Monthly Flight Trend")
plt.xlabel("Month")
plt.ylabel("Number of Flights")
plt.show()
```

```python
#4. Histogram – Departure Delay Distribution
data['DEPARTURE_DELAY'].plot(kind='hist', bins=50, figsize=(10,5))
plt.title("Distribution of Departure Delays")
plt.xlabel("Minutes")
plt.ylabel("Frequency")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
data = pd.read_csv(r"C:\Users\banuk\Downloads\AirLine_Flights_Dataset\flights.csv", low_memory=False)


# Calculate average delays
avg_delays = data[['AIRLINE_DELAY', 'WEATHER_DELAY', 'LATE_AIRCRAFT_DELAY']].mean()

# Bar Chart – Delay Comparison
avg_delays.plot(kind='bar', figsize=(6, 6))

plt.title("Average Delay Comparison by Cause")
plt.xlabel("Delay Type")
plt.ylabel("Average Delay (minutes)")
plt.xticks(rotation=30)
plt.grid(axis='y')
plt.show()

```
