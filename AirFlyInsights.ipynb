Handling Missing Values

```python
num_col = ['price', 'co2_emissions', 'avg_co2_emission_for_this_route']

for col in num_col:
  median_val = df[col].median()
  df[col].fillna(median_val, inplace=True)

cat_col = ['aircraft_type', 'co2_percentage']
for col in cat_col:
  mode_val = df[col].mode()[0]
  df[col].fillna(mode_val, inplace=True)


print("Null values after filling missing:")
print(df.isnull().sum())
```

Feature Engineering

```python
df['duration_hours'] = df['duration'] / 60 # Flight duration in hour

# Extract departure time features
df['departure_hour'] = df['departure_time'].dt.hour
df['departure_day'] = df['departure_time'].dt.day
df['departure_month'] = df['departure_time'].dt.month
df['departure_weekday'] = df['departure_time'].dt.weekday

df.head()
```

A) IQR Method (Interquartile Range)

```python
def remove_outliers_iqr(df, col):
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    return df[(df[col] >= lower) & (df[col] <= upper)]
df = remove_outliers_iqr(df, 'price')
```

```python
sns.boxplot(x=df['price'])
plt.show()
```
